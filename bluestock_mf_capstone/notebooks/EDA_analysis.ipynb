```python
nav = pd.read_csv(
    r"C:\Users\sb231\OneDrive\Desktop\bluestock DA\data\raw data\02_nav_history.csv"
)
```

```python
nav['date'] = pd.to_datetime(nav['date'])
```

```python
nav = nav.sort_values(
    ['amfi_code','date']
)
```

```python
nav = nav.drop_duplicates()
```

```python
nav['nav'] = (
    nav.groupby('amfi_code')['nav']
    .ffill()
)
```

```python
invalid_nav = nav[nav['nav'] <= 0]
```

```python
nav = nav[nav['nav'] > 0]
```

```python
txn = pd.read_csv(
    r"C:\Users\sb231\OneDrive\Desktop\bluestock DA\data\raw data\08_investor_transactions.csv"
)
```

```python
txn['transaction_type'] = (
    txn['transaction_type']
    .str.strip()
    .str.upper()
)
```

```python
mapping = {
    'SYSTEMATIC INVESTMENT':'SIP',
    'SIP':'SIP',
    'LUMPSUM':'LUMPSUM',
    'REDEEM':'REDEMPTION',
    'REDEMPTION':'REDEMPTION'
}

txn['transaction_type'] = txn[
    'transaction_type'
].replace(mapping)
```

```python
print(txn.columns)
```

```python
txn = txn[
    txn['amount_inr'] > 0
]
```

```python
txn['transaction_date'] = pd.to_datetime(
    txn['transaction_date']
)
```

```python
allowed = [
    'Verified',
    'Pending',
    'Rejected'
]

txn = txn[
    txn['kyc_status'].isin(allowed)
]
```

```python
txn.to_csv(
    r"C:\Users\sb231\OneDrive\Desktop\bluestock DA\data\raw data\08_investor_transactions.csv",
    index=False
)
```
