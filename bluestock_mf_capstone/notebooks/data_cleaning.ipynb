```python
import os
import pandas as pd
from sqlalchemy import create_engine
```

```python
BASE_DIR = r"C:\Users\sb231\OneDrive\Desktop\bluestock DA"

processed_folder = os.path.join(BASE_DIR, "data", "processed")
database_folder = os.path.join(BASE_DIR, "database")

os.makedirs(database_folder, exist_ok=True)
```

```python
print(os.listdir(processed_folder))
```

```python
db_path = os.path.join(database_folder, "bluestock_mf.db")

engine = create_engine(f"sqlite:///{db_path}")

print("Database location:", db_path)
```

```python
for file in os.listdir(processed_folder):
    if file.endswith(".csv"):
        path = os.path.join(processed_folder, file)

        df = pd.read_csv(path)

        table_name = file.replace(".csv", "")

        df.to_sql(
            table_name,
            engine,
            if_exists="replace",
            index=False
        )

        print(f"Loaded {table_name} ({len(df)} rows)")
```

```python
from sqlalchemy import inspect

inspector = inspect(engine)

print(inspector.get_table_names())
```

```python
for table in inspector.get_table_names():

    query = f"SELECT COUNT(*) AS rows FROM {table}"

    count = pd.read_sql(query, engine)

    print(table)

    print(count)

    print("-"*40)
```
