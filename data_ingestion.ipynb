```python
import pandas as pd
import numpy as np
import os
import requests
```

```python
folder = "raw data"

files = os.listdir(folder)

datasets = {}

for file in files:
    if file.endswith(".csv"):
        path = os.path.join(folder, file)
        df = pd.read_csv(path)
        datasets[file] = df

        print("=" * 60)
        print(file)
        print("Shape:", df.shape)
        print(df.head())
        print(df.dtypes)
```

```python
codes = [
    119551,
    120503,
    118632,
    119092,
    120841
]

for code in codes:
    url = f"https://api.mfapi.in/mf/{code}"

    response = requests.get(url)

    data = response.json()

    df = pd.DataFrame(data["data"])

    df.to_csv(f"raw data/{code}.csv", index=False)

    print(f"{code} downloaded successfully")
```
